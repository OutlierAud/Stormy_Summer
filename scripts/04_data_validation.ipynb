# Data Validation of The Combined Typhoon & SST Dataset

`data/features/typhoon_sst.csv`

Before conducting any exploratory or statistical analyses, it is important to verify that the processed dataset is complete, consistent, and suitable for analysis. This notebook performs a series of data quality checks on the merged typhoon and sea surface temperature (SST) dataset, including checks for **missing values, duplicate records, data types, and plausible value ranges**. The results provide confidence that the dataset is appropriate for the subsequent spatial exploration and statistical modelling.

```r
#load library
library(tidyverse)

#load dataset
typhoon <- read_csv("../data/features/typhoon_sst.csv")

#dataset dimensions
dim(typhoon)
glimpse(typhoon)
```

#### Check for missing values

```r
# Missing value check
cat(strrep("=",50), "\n")
cat("Missing values by column:\n")
cat(strrep("=",50), "\n")
print(colSums(is.na(typhoon)))

# Duplicate rows check
cat(strrep("=",50), "\n")
cat("No. of duplicate rows: ", sum(duplicated(typhoon)), "\n")
cat(strrep("=",50), "\n")
cat("\nRange of SST:", range(typhoon$sst, na.rm = TRUE), "\n")
cat("Range of Latitude:", range(typhoon$lat, na.rm = TRUE), "\n")
cat("Range of Longitude:", range(typhoon$lon, na.rm = TRUE), "\n")
summary(typhoon)
```
